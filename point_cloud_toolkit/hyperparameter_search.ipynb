## Hyperparameter Search

```python
from dependencies import *
from utils.custom_plotting import LinePlot

import pandas as pd
```

```python
MODEL_ROOT = "D:/models/hyperparameter_search/kc46/"

plot = LinePlot( title = f'Hyperparameter Selection - KC-46 model',
                x_axis_title = 'Epochs',
                y1_axis_title = 'Validation - Segmentation Sparse Categorical Accuracy',
                y2_axis_title = 'Validation Loss' )

best: dict[str, list] = { 'KC-46 Model': [], 'Best Epoch': [], 'Seg Acc': [], 'Loss': [] }
models: dict[str, dict] = {}
for model_dir in os.listdir( MODEL_ROOT ):
    with open( f"{MODEL_ROOT}{model_dir}/final/{model_dir}_final_history.json", "r" ) as j:
        models[model_dir] = json.load( j )

        plot.add_y1_trace( np.array( models[model_dir]['segmentation_output_sparse_categorical_accuracy'] ), model_dir )
        plot.add_y2_trace( np.clip( np.array( models[model_dir]['val_loss'] ), 0, 1000 ), model_dir )

        best['KC-46 Model'].append( model_dir )
        best['Best Epoch'].append( np.argmin( np.array( models[model_dir]['val_loss'] ) ) + 1 )
        best['Seg Acc'].append( np.max( np.array( models[model_dir]['segmentation_output_sparse_categorical_accuracy'] ) ) )
        best['Loss'].append( np.min( np.array( models[model_dir]['val_loss'] ) ) )

best_df = pd.DataFrame( data = best )
display( best_df )
plot.show()
plot.get_fig().write_image( 'E:/AFIT/AAR/5_Thesis/doc/figures/hyperparameter_search/kc-46_full.png', width = 1200, height = 600 )
best_df.to_latex( 'E:/AFIT/AAR/5_Thesis/doc/tables/hyperparameter_search/kc-46_hp_search_results.tex', float_format = '%.3f' )
```
