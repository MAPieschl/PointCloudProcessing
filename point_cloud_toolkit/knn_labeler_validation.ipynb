## kNN Autolabeler Validation

The autolabeler validation tool requires autolabeled samples, along with a manual label corrections. It then computes the F1-score of the autolabeled samples and bootstraps 10,000 additional samples to construct a $95\%$ confidence interval.

```python
import os

import numpy as np
import pandas as pd
```

Randomly sample the available data for validation.

```python
PREDICTION_DIR = 'D:/collect_1_predictions_working/lidar_predictions'
NUM_SAMPLES = 20

np.random.seed( 42 )

available_frames = np.array( os.listdir( PREDICTION_DIR ) )
chosen_frames = available_frames[ np.asarray( np.random.choice( range( 0, len( available_frames ) ), NUM_SAMPLES, replace = False ), dtype = int ) ]

manual_verification = [ 'knn_labels', 'manual_false_pos', 'manual_false_neg' ]
part_labels = [ 'wing', 'fuselage', 'hstab', 'vstab' ]

indicies = [ (outer, inner) for outer in manual_verification for inner in part_labels ]
multicol = pd.MultiIndex.from_tuples( indicies, names = ['Category', 'Part'] )

verification_df = pd.DataFrame( index = chosen_frames, columns = multicol ).fillna( 0 )

for frame in chosen_frames:
    with open( f'{PREDICTION_DIR}/{frame}', 'r' ) as f:
        for l in f:
            l = l.strip()
            start_idx = l.find( ')' )
            knn_label = l[start_idx + 1:].split( ' ' )[-1]

            verification_df.loc[frame, ( 'knn_labels', knn_label )] += 1

display( verification_df )
            
```

```python
precision = lambda tp, fp, fn : tp / ( tp + fp )
recall = lambda tp, fp, fn : tp / ( tp + fn )
```

```python
tp: dict[str, int] = {}
fp: dict[str, int] = {}
fn: dict[str, int] = {}
for p in part_labels:
    fp[p] = verification_df.loc[chosen_frames, ( 'manual_false_pos', p )].sum()
    fn[p] = verification_df.loc[chosen_frames, ( 'manual_false_neg', p )].sum()
    tp[p] = verification_df.loc[chosen_frames, ( 'knn_labels', p )].sum() - fp[p]

f_sample = ( 1 / ( len( part_labels ) ) ) * np.sum( np.array( [ (2 * precision( tp[p], fp[p], fn[p] ) * recall( tp[p], fp[p], fn[p] ) ) / ( precision( tp[p], fp[p], fn[p] ) + recall( tp[p], fp[p], fn[p] ) ) for p in part_labels ] ) )
```

```python
f_bs = []
np.random.seed( None )
for sm in range( 10000 ):
    sampled_frames = chosen_frames[ np.asarray( np.random.choice( range( 0, len( chosen_frames ) ), NUM_SAMPLES, replace = True ), dtype = int ) ]

    tp: dict[str, int] = {}
    fp: dict[str, int] = {}
    fn: dict[str, int] = {}
    for p in part_labels:
        fp[p] = verification_df.loc[sampled_frames, ( 'manual_false_pos', p )].sum()
        fn[p] = verification_df.loc[sampled_frames, ( 'manual_false_neg', p )].sum()
        tp[p] = verification_df.loc[sampled_frames, ( 'knn_labels', p )].sum() - fp[p]

    f_bs.append( ( 1 / ( len( part_labels ) ) ) * np.sum( np.array( [ (2 * precision( tp[p], fp[p], fn[p] ) * recall( tp[p], fp[p], fn[p] ) ) / ( precision( tp[p], fp[p], fn[p] ) + recall( tp[p], fp[p], fn[p] ) ) for p in part_labels ] ) ) )

print( f"The bootstrapped 95% confidence interval is ({np.percentile( np.array( f_bs ), 2.5 ):.3f}, {np.percentile( np.array( f_bs ), 97.5 ):.3f})" )
```
