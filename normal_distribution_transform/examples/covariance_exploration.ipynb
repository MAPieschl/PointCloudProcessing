```python
import numpy as np
```

```python
small_cloud = np.random.normal( 0, 1, size = (10, 3) )
large_cloud = np.random.normal( 0, 1, size = (100000, 3) )

sig_small = np.cov( small_cloud.T )
sig_large = np.cov( large_cloud.T )

det_small = np.linalg.det( sig_small )
det_large = np.linalg.det( sig_large )

print( sig_small )

print( det_small )
print( det_large )
```
